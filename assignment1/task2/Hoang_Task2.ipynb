```python
import cv2
import matplotlib.pyplot as plt

root = Path('assignment1/task2')
items = [
    ('detailed_noisy', 'Nhiều chi tiết + nhiễu'),
    ('smooth_gradient', 'Gradient mượt'),
    ('high_contrast', 'Tương phản cao'),
]
fig, axes = plt.subplots(3, 3, figsize=(12, 10))
for row, (name, label) in enumerate(items):
    paths = [root/'images'/f'{name}.png', root/'results'/f'{name}_mean_k5.png', root/'results'/f'{name}_gaussian_k5_sigma1.2.png']
    for col, (path, title) in enumerate(zip(paths, ['Original', 'Mean 5×5', 'Gaussian 5×5, σ=1.2'])):
        axes[row, col].imshow(cv2.imread(str(path), cv2.IMREAD_GRAYSCALE), cmap='gray', vmin=0, vmax=255)
        axes[row, col].set_title(f'{label} — {title}')
        axes[row, col].axis('off')
plt.tight_layout()
```

## Nhận xét ngắn

- Mean Filter lấy trung bình đều trong vùng 5×5 nên làm mờ nhiễu và biên rõ hơn.
- Gaussian Filter ưu tiên pixel gần tâm, thường giữ chuyển tiếp tự nhiên hơn Mean Filter.
- Kernel và sigma lớn hơn sẽ làm mượt mạnh hơn, đồng thời mất thêm chi tiết.
